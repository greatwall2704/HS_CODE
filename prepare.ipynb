```python
import pandas as pd  
import json
import os
from sklearn.model_selection import train_test_split

```

```python
# Đọc file CSV
df = pd.read_csv('data/QA data/QAdataset.csv')
df.info()
```

```python
# Tách DataFrame thành 3 phần: train (70%), val (20%), test (10%)
train_df, temp_df = train_test_split(df, test_size=0.3, random_state=42)  # 70% train, 30% temp
val_df, test_df = train_test_split(temp_df, test_size=0.33, random_state=42)  # 20% val, 10% test

print(f'Train: {len(train_df)} mẫu ({len(train_df)/len(df)*100:.1f}%)')
print(f'Val: {len(val_df)} mẫu ({len(val_df)/len(df)*100:.1f}%)')
print(f'Test: {len(test_df)} mẫu ({len(test_df)/len(df)*100:.1f}%)')
print(f'Total: {len(df)} mẫu')
```

```python
# Lưu train dataset
train_output_path = 'vit5-chatbot-finetune/data/train_dataset.json'
train_data = []
for _, row in train_df.iterrows():
    train_data.append({
        'mahs': row['mahs'],
        'prompt': row['prompt'], 
        'response': row['response']
    })

with open(train_output_path, 'w', encoding='utf-8') as f:
    json.dump(train_data, f, ensure_ascii=False, indent=2)

# Lưu validation dataset  
val_output_path = 'vit5-chatbot-finetune/data/val_dataset.json'
val_data = []
for _, row in val_df.iterrows():
    val_data.append({
        'mahs': row['mahs'],
        'prompt': row['prompt'],
        'response': row['response']
    })

with open(val_output_path, 'w', encoding='utf-8') as f:
    json.dump(val_data, f, ensure_ascii=False, indent=2)

# Lưu test dataset
test_output_path = 'vit5-chatbot-finetune/data/test_dataset.json'
test_data = []
for _, row in test_df.iterrows():
    test_data.append({
        'mahs': row['mahs'],
        'prompt': row['prompt'],
        'response': row['response']
    })

with open(test_output_path, 'w', encoding='utf-8') as f:
    json.dump(test_data, f, ensure_ascii=False, indent=2)

print(f'Đã lưu:')
print(f'- Train: {train_output_path}')
print(f'- Val: {val_output_path}')
print(f'- Test: {test_output_path}')
```
